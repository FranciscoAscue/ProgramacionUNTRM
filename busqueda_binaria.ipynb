````python
def busqueda_binaria(A,n):
  inicio = 0
  fin = len(A) - 1
  while inicio < fin:
    medio = (inicio + fin) // 2

    if A[medio] == n:
      return medio
    elif A[medio] < n:
      inicio = medio +1
    else:
      fin = medio

  return -1
````

````python
def busqueda_lineal(A,n):
  for i in range(len(A)):
    if A[i] == n:
      return i
  return -1
````

## Recursividad

$$f(x) = f(x-1) + 2$$

$$\begin{cases}
0 & \text{si } x = 0 \\
1 & \text{si } x = 1 \\
f(x-1) + 2  & \text{si } x > 1
\end{cases}$$

### Fibonacci

$$f(x) = f(x-1) + f(x-2)$$

$$\begin{cases}
0 & \text{si } x = 0 \\
1 & \text{si } x = 1 \\
f(x-1) + f(x-2)  & \text{si } x > 1
\end{cases}$$


````python
## Recursividad

def recursivo(x):
  if x == 0:
    return 0
  elif x == 1:
    return 1
  return recursivo(x-1) + 2

def fibonacci(n):
  if n == 0:
    return 0
  elif n == 1:
    return 1

  return fibonacci(n-1) + fibonacci(n-2)
````

````python
def busqueda_binaria_recursiva(Arr,n):
  if not Arr:
    return -1

  medio = len(Arr) // 2

  if Arr[medio] == n:
    return medio

  if Arr[medio] > n:
    return busqueda_binaria_recursiva(Arr[:medio], n)

  res = busqueda_binaria_recursiva(Arr[medio+1:]),n

  if res == -1:
    return -1
````
